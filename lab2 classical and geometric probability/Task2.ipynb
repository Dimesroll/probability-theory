## Задача 1.38: Перестановки книг на полке

**Условие:**  
Пятитомное собрание сочинений расположено на полке в случайном порядке. Какова вероятность того, что книги стоят слева направо в порядке нумерации томов (от 1 до 5)?

**Математическая модель:**
* Общее число способов расставить $N=5$ книг равно количеству перестановок $N!$:
  $$n = 5! = 120$$
* Благоприятный исход только 1 — строго определенный порядок $1, 2, 3, 4, 5$:
  $$m = 1$$
* По классическому определению вероятности:
  $$P = \frac{m}{n} = \frac{1}{120}$$

```python
n = math.factorial(5)
p  = 1 / n

print(f"Общее количество способов расстановки: {n}")
print(f"Вероятность: P = 1 / 120 = {p:.6f}")
```

## Задача 1.43: Анализ пятизначного телефонного номера

**Условие:**  
Наудачу взятый телефонный номер состоит из 5 цифр. Какова вероятность того, что в нем все цифры:  
а) различные; б) одинаковые; в) нечетные?  
*Известно, что номер телефона не начинается с цифры 0.*

**Математическая модель:**
Первая цифра из 9 возможных (1-9), остальные 4 цифры из 10 (0-9).  
Общее число номеров: $n = 9 \cdot 10^4 = 90\,000$.

* **а) Все различные:** $9 \cdot 9 \cdot 8 \cdot 7 \cdot 6$ вариантов.
* **б) Все одинаковые:** $11111, 22222, \dots, 99999$ (всего 9 вариантов).
* **в) Все нечетные:** каждая из 5 позиций выбирается из 5 нечетных цифр ($1, 3, 5, 7, 9$), т.е. $5^5 = 3125$ вариантов.

```python
total = 9 * (10**4)

# а) Все разные
fav_a = 9 * 9 * 8 * 7 * 6
p_a = fav_a / total

# б) Все одинаковые
fav_b = 9
p_b = fav_b / total

# в) Все нечетные
fav_v = 5**5
p_v = fav_v / total

print(f"[Задача 1.43]")
print(f"а) Все цифры различные: P = {fav_a}/{total} = {p_a:.4f}")
print(f"б) Все цифры одинаковые: P = {fav_b}/{total} = {p_b:.6f}")
print(f"в) Все цифры нечетные:  P = {fav_v}/{total} = {p_v:.4f}")
```

## Задача 1.48: Игра в 3 кости (Сумма очков)

**Условие:**  
При бросании трех игральных костей найти вероятности:  
а) выпадения суммы в 11 очков;  
б) выигрыша (сумма очков превосходит 10).

**Математическая модель:**
Общее число равновероятных исходов при броске трех костей: $n = 6^3 = 216$.
* **Пункт а):** Точный подсчет всех троек $(d_1, d_2, d_3)$, у которых $d_1 + d_2 + d_3 = 11$. Дает $27$ комбинаций.
* **Пункт б):** Из-за симметрии распределения сумм от $3$ до $18$, ровно половина исходов ($108$) дает сумму $\le 10$, и ровно половина ($108$) — сумму $> 10$.
  $$P(\text{выигрыша}) = \frac{108}{216} = 0.5$$

```python
comb = 0
comb_win = 0

# Алгоритмический перебор всех вариантов выпадения 3 кубиков
for d1 in range(1, 7):
    for d2 in range(1, 7):
        for d3 in range(1, 7):
            s = d1 + d2 + d3
            if s == 11:
                comb += 1
            if s > 10:
                comb_win += 1

total = 6 ** 3
p_a = comb / total
p_b = comb / total

print(f"а) Количеств комбинаций с суммой 11: {comb}: P = {comb}/{total} = {p_a:.4f}")
print(f"б) Количеств комбинаций с суммой > 10: {comb_win}: P = {comb_win}/{total} = {p_b:.4f}")
```
