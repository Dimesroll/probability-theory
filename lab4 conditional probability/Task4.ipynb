# Практическое задание: Формула полной вероятности

## Условие задачи

В одной урне находится 15 белых и 5 черных шаров, а во другой — 6 белых и 1 черный шар. 
Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. 

**Задание:**
1. Найти вероятность того, что шар, вынутый из второй урны, окажется белым (классический математический расчет).
2. Смоделировать процесс проведения данного эксперимента методом Монте-Карло на Python.

---

### Классическое решение

Пусть $A$ — событие «из второй урны вынут белый шар»; $H_k$ — гипотеза «из первой во вторую урну перенесено ровно $k$ белых шаров», $k = 0, 1, 2, 3$.

Всего способов выбрать 3 шара из первой урны: $C_{20}^3 = 1140$.

Вероятность гипотезы $H_k$ (гипергеометрическое распределение):
$$P(H_k) = \frac{C_{15}^k C_5^{3-k}}{C_{20}^3}.$$

После переноса во второй урне будет 10 шаров, среди которых $6 + k$ белых. Поэтому
$$P(A \mid H_k) = \frac{6 + k}{10}.$$

Вычисляем используя формулу полной вероятности:
$$P(A) = \sum_{k=0}^{3} P(H_k) P(A \mid H_k)$$

---

```python
#Классическое решение
import math

total_combinations = math.comb(20, 3)
p_a = 0.0

print(f"Общее количество способов C(20, 3) = {total_combinations}\n")
print("Расчёт вероятностей гипотез и условных вероятностей:")

for k in range(4):
    p_hk = (math.comb(15, k) * math.comb(5, 3 - k)) / total_combinations    
    p_a_given_hk = (6 + k) / 10
    p_a += p_hk * p_a_given_hk
    
    print(f"k = {k}: P(H_{k}) = {p_hk:.6f}, P(A | H_{k}) = {p_a_given_hk:.2f}")

print(f"Точная вероятность P(A) = {p_a:.4f}")
```

```python
import random

rng = random.Random(20260928)
checkpoints = (100, 1_000, 10_000, 100_000)
white_outcomes = 0
estimates = []

for experiment in range(1, checkpoints[-1] + 1):
    first_urn = ['белый'] * 15 + ['чёрный'] * 5
    second_urn = ['белый'] * 6 + ['чёрный']

    transferred = rng.sample(first_urn, 3)  # без возвращения
    second_urn.extend(transferred)
    drawn = rng.choice(second_urn)
    if drawn == 'белый':
        white_outcomes += 1

    if experiment in checkpoints:
        estimate = white_outcomes / experiment
        estimates.append(estimate)
        print(f'{experiment:>6} испытаний: P ≈ {estimate:.5f}; '
              f'отклонение = {abs(estimate - float(probability)):.5f}')

assert abs(estimates[-1] - float(probability)) < 0.01
```
